####  13.	List	the	name,salary,comm.	For	those	employees	whose	net	pay	is greater	than	or	equal	to	any	other	employee	salary	of	the	company.

```python
%sql

-- Employees whose net pay >= ANY other employee salary
WITH emp_net AS (
  SELECT
    empno,
    ename,
    sal,
    COALESCE(comm, 0) AS comm,
    COALESCE(sal, 0) + COALESCE(comm, 0) AS net_pay
  FROM Emp
)
SELECT
  e.ename,
  e.sal,
  e.comm,
  e.net_pay
FROM emp_net e
WHERE e.net_pay >= (
  SELECT MIN(sal)
  FROM Emp
  WHERE empno <> e.empno
)
ORDER BY e.net_pay DESC, e.ename;
```

#### 38. List	the	name	of	the	dept	where	more	than	average	no.	of	emps	are working.

```python
%sql

-- Databricks SQL
WITH dept_counts AS (
  SELECT
    deptno,
    COUNT(*) AS emp_cnt
  FROM Emp
  GROUP BY deptno
),
avg_cnt AS (
  SELECT AVG(emp_cnt) AS avg_emp_cnt
  FROM dept_counts
)
SELECT
  d.dname
FROM dept_counts AS c
JOIN Dept AS d
  ON d.deptno = c.deptno
CROSS JOIN avg_cnt
WHERE c.emp_cnt > avg_emp_cnt
GROUP BY d.dname;
```

#### 39. List	the	Managers	name	who	is	having	max	no.of	emps	working	under him.

```python

%sql
WITH mgr_counts AS (
  SELECT
    m.empno        AS mgr_empno,
    m.ename        AS manager_name,
    COUNT(e.empno) AS report_count
  FROM Emp e
  JOIN Emp m
    ON e.mgr = m.empno
  GROUP BY m.empno, m.ename
)
SELECT manager_name, report_count
FROM mgr_counts
WHERE report_count = (
  SELECT MAX(report_count) FROM mgr_counts
);
```

```python
%sql
SELECT m.ename AS manager_name,
       COUNT(e.empno) AS report_count
FROM Emp e
JOIN Emp m
  ON e.mgr = m.empno
GROUP BY m.ename
HAVING COUNT(e.empno) = (
    SELECT MAX(cnt)
    FROM (
        SELECT COUNT(*) AS cnt
        FROM Emp
        WHERE mgr IS NOT NULL
        GROUP BY mgr
    ) t
);
```
