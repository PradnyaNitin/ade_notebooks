#### 9.7-DataFrame Read Modes
* Permissive
* DropMalformed
* FailFast


```python
# Step 1- Read the data first (bad_data.csv)
# dbutils.fs.rm("/Volumes/workspace/default/upskill_volume/bad_data.csv", recurse=True)

df = spark.read \
    .option("header", True) \
    .option("mode", "PERMISSIVE") \
    .option("columnNameOfCorruptRecord", "_corrupt_record") \
    .csv("/Volumes/workspace/default/practice_volume/bad_data.csv")

display(df)
```

#### 9.7.1- Permissive

```python
# 1️⃣ PERMISSIVE (default) — keeps bad rows in a special column (_corrupt_record)
df_permissive = (
    spark.read
        .option("header", "true")
        .option("mode", "PERMISSIVE")
        .option("inferSchema", "true")
        .csv('/Volumes/workspace/default/practice_volume/bad_data.csv')
)

print("==== PERMISSIVE MODE ====")
display(df_permissive)
df_permissive.printSchema()
```

#### 9.7.2- DropMalformed

```python
# 2️⃣ DROPMALFORMED — drops rows that do not match schema
df_drop = (
    spark.read
        .option("header", "true")
        .option("mode", "DROPMALFORMED")
        .option("inferSchema", "true")
        .csv("/Volumes/workspace/default/practice_volume/bad_data.csv")
)

print("==== DROPMALFORMED MODE ====")
display(df_drop)
df_drop.printSchema()
```

#### 9.7.3- FailFast

```python
# 2️⃣ DROPMALFORMED — drops rows that do not match schema
df_failfast = (
    spark.read
        .option("header", "true")
        .option("mode", "FAILFAST")
        .option("inferSchema", "true")
        .csv('/Volumes/workspace/default/practice_volume/bad_data.csv')
)

print("==== FAILFAST MODE ====")
display(df_failfast)
df_drop.printSchema()
```

```python
# 3️⃣ FAILFAST — stops immediately when a bad row is found
try:
    df_fail = (
        spark.read
            .option("header", "true")
            .option("mode", "FAILFAST")
            .option("inferSchema", "true")
            .csv('/Volumes/workspace/default/practice_volume/bad_data.csv')
    )
    
    print("==== FAILFAST MODE ====")
    df_fail.show(truncate=False)

except Exception as e:
    print("FAILFAST MODE ERROR:")
    print(e)
```

#### The default mode is Permissive
* What does PERMISSIVE do?
* It does not fail when it encounters bad/malformed data.
* It fills missing columns with null.



#### 9.8-Dataframe Write Modes


* 1-Overwrite-              Replaces existing data at the path/table
* 2-Append-                 Adds data to existing file/table (same schema required)
* 3-Ignore-                 Skips writing if data already exists (does nothing)   
* 4-error or errorifexists- Throws error if destination exists     


```python
# Step 1- Create 1st Dataframe
columns = ["id", "name", "salary"]

data1 = [
    (1, "Alice", 70000),
    (2, "Bob", 80000),
    (3, "Charlie", 75000)
]

df1 = spark.createDataFrame(data=data1, schema=columns)
display(df1)
```

#### 1. overwrite

```python
# Step 2- Write First Time using (overwrite)
df1.write.mode("overwrite").csv("/Volumes/workspace/default/practice_volume/write_demo/")
# This creates the folder and writes the first 3 rows.
```

```python
# Step 3- Create 2nd Dataframe with additional records, Because DataFrames are immutable, we cannot modify df1
columns = ["id", "name", "salary"]
data2 = [
    (4, "David", 72000),
    (5, "Eva", 68000),
    (6, "Frank", 79000)
]

df2 = spark.createDataFrame(data=data2, schema=columns)
display(df2)
```

#### 2. append

```python
# Step 4- APPEND Mode Demo, Append df2 to the existing folder:
df2.write.mode("append").csv("/Volumes/workspace/default/practice_volume/write_demo/")
```

```python
# Step 5- Read back combined data
df_read = spark.read.csv(
    "/Volumes/workspace/default/practice_volume/write_demo/",
    header=True,
    inferSchema=True
)

df_read.show()
```

```python
# Step 6- STEP 6 — OVERWRITE Mode Demo
columns = ["id", "name", "salary"]
data3 = [
    (10, "Zara", 90000)
]

df3 = spark.createDataFrame(data3, columns)
df3.show()

# Overwrite the folder:
df3.write.mode("overwrite").csv("/Volumes/workspace/default/practice_volume/write_demo/")

# Read Again to Show Overwrite Effect
df_read2 = spark.read.csv(
    "/Volumes/workspace/default/practice_volume/write_demo/",
    header=False,
    inferSchema=True
)

df_read2.show()
```

#### 3. ignore
* Use case: Write only if folder is NOT present
* If folder exists, skip writing
* No error, no overwrite
* Safe for accidental writes

```python
df3.write.mode("ignore").csv("/Volumes/workspace/default/practice_volume/write_demo/")
```

#### 4. error or errorifexists (Default)
* Throw error if folder exists, Error: Path already exists

#### 9.9-Schema Management

##### Ways to Handle Schema in PySpark
* 1- Infer Schema Automatically (using .option("inferSchema", True))


* 2- Define Schema Manually (using StructType) 

```python
# 1- Infer Schema from data directly
df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/workspace/default/practice_volume/Employees.csv")
display(df)
df.printSchema()

# Spark looks at the data and tries to guess the data types.

```

```python
# 2- Define Schema Manually (using StructType)

from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType

schema = StructType([
    StructField("ID", IntegerType(), True),
    StructField("FIRST_NAME", StringType(), True),
    StructField("LAST_NAME", StringType(), True),
    StructField("DEPARTMENT", StringType(), True),
    StructField("SALARY", FloatType(), True)
])

df = spark.read \
    .option("header", True) \
    .schema(schema) \
    .csv("/Volumes/workspace/default/practice_volume/Employees.csv")
display(df)
df.printSchema()

# More reliable and controlled.
```

#### 9.10-Challenges of schema inference
* **Challenges of Schema Inference**
* 1. Incorrect type guessing
* Example: CSV file:
* id,name,age
* 1,Alice,25
* 2,Bob,thirty

* **2. Null values confuse inference**
* If your column has only nulls in the first few rows, Spark might infer it as NullType, which causes errors in transformations.

* **3. Slower for big data**
* When Spark infers schema, it scans part of the data to guess types. For huge files, this:
* 1- Takes more time
* 2- Might give wrong schema if early rows don’t represent the full file


#### 9.11-Introduction to schema enforcement
* **1-Schema DDL**
* **2-Struct type**

* **Q-What is Schema Enforcement?**
* Schema enforcement means you tell Spark exactly what the structure of the data should be (column names + data types), instead of letting it guess.

* **This avoids issues like:**
•	Wrong data types
•	Column mismatch
•	Null confusion
•	Performance delays during schema inference


#### 9.11.1. Schema Using DDL String (Declarative schema)
* DDL = Data Definition Language
* Just like SQL: "column_name data_type, ..."


```python
schema_ddl = "ID INT, FIRST_NAME STRING, LAST_NAME STRING, DEPARTMENT STRING, SALARY DOUBLE"

df = spark.read \
    .option("header", True) \
    .schema(schema_ddl) \
    .csv("/Volumes/workspace/default/practice_volume/Employees.csv")

df.printSchema()
df.show()
```

#### 9.11.2-Schema Using StructType (Programatic schema)
* This is the programmatic way of defining schema using pyspark.sql.types.


```python
# 2- Define Schema Manually (using StructType)

from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType

schema = StructType([
    StructField("ID", IntegerType(), True),
    StructField("FIRST_NAME", StringType(), True),
    StructField("LAST_NAME", StringType(), True),
    StructField("DEPARTMENT", StringType(), True),
    StructField("SALARY", FloatType(), True)
])

df = spark.read \
    .option("header", True) \
    .schema(schema) \
    .csv("/Volumes/workspace/default/practice_volume/Employees.csv")
display(df)
df.printSchema()

# More reliable and controlled.
```
