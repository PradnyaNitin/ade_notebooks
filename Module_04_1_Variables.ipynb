#### Question 1- Creating Python Variables

#### A variable is a container for storing data values. It's like a label that you can use to refer to a value in your program. You can assign any type of data to a variable (e.g., numbers, text, lists) without declaring its type.

* **Assigning values to variables**
* name = "Alice"          # String
* age = 25                # Integer
* height = 5.6            # Float
* is_student = True       # Boolean

* **Dynamic Typing:** You don’t need to specify the type of the variable (Python figures it out)

* x = 10        # x is an integer
* x = "hello"   # Now x is a string

#### Question 2- Getting Type of a Variable

```python
x = "Zara"
y =  10
z =  10.10

print( "x =", x )
print( "y =", y )
print( "z =", z )
print(type(x))
print(type(y))
print(type(z))
```

#### Question 3- Casting Python Variables

```python
## You can specify the data type of a variable with the help of casting as follows:
x = str(10)    # x will be '10'
y = int(10)    # y will be 10 
z = float(10)  # z will be 10.0

print(x ,y,z)
```

#### Question 4- Case-Sensitivity of Python Variables

```python
## Python variables are case sensitive which means Age and age are two different variables
age = 20
Age = 30

print( "age =", age )
print( "Age =", Age )
```

#### Question 5- Python Variables - Multiple Assignment

```python
a=b=c=10
print (a,b,c)

a,b,c = 10,20,30
print (a,b,c)
```

#### 4.6-DataTypes

```python
•	Numeric - int, float, complex
•	Sequence Type - string, list, tuple
•	Mapping Type - dict
•	Boolean - bool
•	Set Type - set, frozenset
•	Binary Types - bytes, bytearray, memoryview
```

```python
##  Example				            Data Type	
x = "Hello World"			        #str	
print(x)
x = 20					            #int	

x = 20.5					       # float	
x = 1j					            #complex	
print(x)
x = ["apple", "banana", "cherry"]	#list	
print("list = ", x)
x = ("apple", "banana", "cherry")	#tuple	
print(x)
x = range(6)				       # range	
print(x)
x = {"name" : "John", "age" : 36}	#dict	
print(x)
x = {"apple", "banana", "cherry"}	#set	
print(x)
x = True					       # bool	
print(x)
```

#####Summary of Methods Used:
* len(): Returns the length of the string.
* upper(): Converts all characters in the string to uppercase.
* lower(): Converts all characters in the string to lowercase.
* capitalize(): Capitalizes the first character of the string.
* title(): Capitalizes the first character of each word in the string.

#####  Print the length of the string:

```python
str1 = 'Python programming'
print("Length of the string:", len(str1))
```

##### - Print the string in uppercase:

```python
str1 = 'Python programming'
print("String in uppercase:", str1.upper())
```

##### - Print the string in lowercase:

```python
str1 = 'Python programming'
print("String in lowercase:", str1.lower())
```

##### - Print the string with the first letter in capital:

```python
str1 = 'Python programming'
print("String with first letter in capital:", str1.capitalize())
```

##### 1.1 Print the string with Title Format:

##### -Print the result of addition, subtraction, multiplication, division, and modulus operations using these variables.

```python
a = 15
b = 4

# Addition
addition = a + b
print("Addition:", addition)

# Subtraction
subtraction = a - b
print("Subtraction:", subtraction)

# Multiplication
multiplication = a * b
print("Multiplication:", multiplication)

# Division
division = a / b
print("Division:", division)

# Modulus (remainder)
modulus = a % b
print("Modulus:", modulus)
```
