#### 1- Read the csv file

```python
df = spark.read.option('header', 'true')\
    .option('inferSchema', 'true')\
    .csv('/Volumes/workspace/default/practice_volume/Adventureworks_Customers.csv')
display(df)
df.printSchema()
```

#### 10.1-Data Transformation
* 1- Adding new columns (withColumn)
* 2- Renaming columns
* 3- Dropping columns
* 4- Derived columns


##### 1- Adding New Column (withColumn)

```python
# 1- Adding New Column (withColumn)
from pyspark.sql.functions import lit

df = df.withColumn("Country", lit("USA"))
display(df)
```

##### 2- Renaming Columns

```python
# 2- Renaming Columns
df1 = df.withColumnRenamed("AnnualIncome", "Income")
display(df1)
```

#####  3- Dropping Columns

```python
# 3- Dropping Columns
df2 = df1.drop('TotalChildren')
display(df2)
```

##### 4- Derived Column (Full Name)

```python
from pyspark.sql.functions import concat_ws

df3 = df2.withColumn("FullName", concat_ws(" ", df.FirstName, df.LastName))
display(df3)
```

##### 10.3.3- Remove comma(,) from AnnualIncome column

```python
from pyspark.sql.functions import regexp_replace, col
df_no_comma = df.withColumn("NoComma", regexp_replace(col("AnnualIncome"), ",", ""))
display(df_no_comma)
```

##### 10.3.4- Remove dollar $ only:

```python
from pyspark.sql.functions import regexp_replace, col
df_no_dollar = df.withColumn("NoDollar", regexp_replace(col("AnnualIncome"), "\\$", ""))
display(df_no_dollar)

# The dollar sign ($) is a special character that normally means "end of string".

# To treat it literally as a character, you need to escape it: \$

# But in Python (and PySpark), strings also use backslashes for escaping, so to pass a single backslash to the regex engine, you write \\ in the Python string.

```

##### 10.3.5- Remove comma(,) and Doller($) in one code

```python
from pyspark.sql.functions import regexp_replace, col

df_clean = df.withColumn("AnnualIncomeClean",
    regexp_replace(regexp_replace(col("AnnualIncome"), "\\$", ""), ",", "").cast("int")
)
display(df_clean)
```

##### 10.4.3- Union and unionAll

```python
columns = ["id", "name", "salary"]

data1 = [
    (1, "Alice", 70000),
    (2, "Bob", 80000),
    (3, "Charlie", 75000)
]

df1 = spark.createDataFrame(data1, columns)
df1.show()

data2 = [
    (3, "Charlie", 75000),
    (4, "David", 72000),
    (4, "David", 72000),
    (5, "Eva", 68000)
]

df2 = spark.createDataFrame(data2, columns)
df2.show()
```

```python
df_union = df1.union(df2)
df_union.show()
# union showing duplicates

# use distinct to remove duplicate
df_distinct = df1.union(df2).distinct()
df_distinct.show()
```

```python
df_union_all = df1.unionAll(df2)
df_union_all.show()
```

#### 10.5-String and Date Manipulation

#####  10.5.1- Handling date manipulation in Spark
* my date in BirthDate column is like 1990-04-08 which is year-month-day but i want date to print like day-month-year, can you help me with query

```python
from pyspark.sql.functions import to_date, date_format

# Ensure BirthDate is treated as a date type (if not already)
df = df.withColumn("BirthDate", to_date("BirthDate", "yyyy-MM-dd"))

# Format BirthDate as 'dd-MM-yyyy'
df = df.withColumn("FormattedBirthDate", date_format("BirthDate", "dd-MM-yyyy"))

# Show result
df.select("BirthDate", "FormattedBirthDate").show(truncate=False)
```

```python
from pyspark.sql.functions import to_date, year, month

df2 = df.withColumn("BirthYear", year("BirthDate")).withColumn("BirthMonth", month("BirthDate"))
df2.select('BirthYear', 'BirthMonth').show()
```

##### 10.5.2- Handling string manipulation in Spark


```python
from pyspark.sql.functions import upper, lower

df_string = df.withColumn("FirstNameUpper", lower("FirstName"))
display(df_string.select('FirstNameUpper', 'FirstName'))
```

##### 10.5.3- Handling timestamp manipulation in Spark

```python
from pyspark.sql.functions import dayofmonth

df_time = df.withColumn("BirthDay", dayofmonth("BirthDate"))
display(df_time)
```

#### 10.6- Types of joins:
* **Syntax- dataframe1.join(dataframe2,dataframe1.column_name ==  dataframe2.column_name,"type")** 

* 1- Left join
* 2- Right join
* 3- Full outer join
* 4- Inner join
* 5- Left anti join
* 6- Right anti join

```python
# 1st Dataframe
data = ([1, "sravan", "TCS"],
        [2, "ojaswi", "WIPRO"], 
        [3, "rohith", "CAPGEMINI"],
        [4, "sridevi", "COGNIZANT"], 
        [5, "bobby", "ZENSAR"])

cols = ['ID', 'Name', 'Company']
employee = spark.createDataFrame(data, cols)
display(employee)
```

```python
# 2nd Dataframe
data = ([1, "45000", "IT"],
         [2, "145000", "Manager"],
         [3, "45000", "HR"],
         [4, "34000", "Sales"])

cols = ['ID', 'salary', 'department']
department = spark.createDataFrame(data, cols)

display(department)
```

##### 10.6.1- Left join


```python
employee.join(department, employee.ID == department.ID, "left").show()
```

#####  10.6.2- Right join


```python
employee.join(department, employee.ID == department.ID, "right").show()
```

##### 10.6.3- Full outer join


```python
employee.join(department, employee.ID == department.ID, "fullouter").show()
```

#####  10.6.4- Inner join
* It combines columns from both left and right where the join condition matches.
* Output includes columns from both sides.

```python
employee.join(department, employee.ID == department.ID, "inner").show()
```

#####  10.6.5- Left anti join


```python
employee.join(department, employee.ID == department.ID, "left_anti").show()
```

##### 10.6.6- Left-semi join
* Returns only the left side rows that have a match in right, but does not include any columns from the right side.
* Output includes columns from left only.
* Use case- Use left_semi when you only care whether a match exists in the right DataFrame (like using EXISTS in SQL), but don’t need right’s data.

```python
employee.join(department, employee.ID == department.ID, "left_semi").show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.getOrCreate()

# Left DataFrame (df_left)
data_left = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "David")
]
df_left = spark.createDataFrame(data_left, ["id", "name"])

# Right DataFrame (df_right)
data_right = [
    (2, "Sales"),
    (3, "IT"),
    (5, "HR")
]
df_right = spark.createDataFrame(data_right, ["id", "dept"])

# Show both
df_left.show()
df_right.show()
```

```python
# Left Semi Join (only matching rows from df_left)
df_left_semi = df_left.join(df_right, on="id", how="left_semi")
df_left_semi.show()
```

```python
# Left Anti Join (only non-matching rows from df_left)
df_left_anti = df_left.join(df_right, on="id", how="left_anti")
df_left_anti.show()
```

#### 10.7-Displaying Data

* 2- Limit/take


##### 10.7.1- Show vs display 
* **1-Show**
* Displays DataFrame output in text format in the notebook cell.

* Shows default 20 rows (unless you specify more).

* Truncates long strings to 20 characters (unless truncate=False is passed).

* **Syntax**
* df.show()                       # shows 20 rows, default truncation
* df.show(50, truncate=False)     # shows 50 rows with full column values

* **2-Display**
* Renders output in a rich, interactive table inside Databricks notebooks



##### 10.7.2- 2- Limit/take
* **1- limit**
* Returns a new DataFrame containing the first n rows.
* Lazy evaluation → not executed until an action is called (like show() or collect()).
* limit return dataframe

**Syntax**
* df_limit = df.limit(5)
* df_limit.show()

* **2-Take**
* Immediately executes and returns n rows as a Python list of Row objects
* Eager evaluation (action)
* It return python list([Row,Row..])

**Syntax**
* rows = df.take(5)
