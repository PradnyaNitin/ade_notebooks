```sql
--  [Easy] List the number of employees in each department where the number is more than 3.

select count(*) as total_emp, deptno 
from emp
group by deptno
having total_emp > 3
order by deptno;
```

```sql
--  [Easy] List the names of departments where at least 3 employees are working.
select d.dname, count(*) as total_emp 
from emp e 
join dept d 
on e.deptno = d.deptno
group by d.dname
having total_emp >= 3
```

```sql
--  [Easy] List the employee names and their average salary department-wise.

select e.ename, e.deptno, t.avg_sal
from emp e 
join (
  select deptno, avg(sal) as avg_sal
  from emp
  group by deptno
) t
on e.deptno = t.deptno;

```

```sql
--  [Hard] List the names of employees who are getting the highest salary department-wise.

select max(sal) as highes_sal, deptno
from emp
group by deptno;

select e.ename, e.deptno, highes_sal
from emp e
join (
  select max(sal) as highes_sal, deptno
  from emp
  group by deptno
) t
on e.deptno = t.deptno
and e.sal = t.highes_sal;
```

```sql
--   [Hard] List the name of the department where the highest number of employees are working.
SELECT d.dname, COUNT(*) AS total_emp
FROM emp e
JOIN dept d
  ON e.deptno = d.deptno
GROUP BY d.deptno, d.dname
ORDER BY total_emp DESC
LIMIT 1;
```
