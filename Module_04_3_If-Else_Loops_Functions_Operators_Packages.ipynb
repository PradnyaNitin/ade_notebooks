```python
def display_list(my_list):
    print("Current List:", my_list)

def add_element(my_list, element):
    my_list.append(element)   # append element
    print(f"After adding '{element}':", my_list)

def remove_element(my_list, element):
    if element in my_list:
        my_list.remove(element)
        print(f"After removing '{element}':", my_list)
    else:
        print(f"'{element}' not found in the list")

def update_element(my_list, old, new):
    if old in my_list:
        index = my_list.index(old)
        my_list[index] = new
        print(f"Updated '{old}' to '{new}':", my_list)
    else:
        print(f"'{old}' not found in the list")


# ---------- Using the Functions ----------
my_list = [10, 20, 30, "vijay", "ajay"]

display_list(my_list)
add_element(my_list, "suresh")
remove_element(my_list, 20)
update_element(my_list, "vijay", "Ramesh")
```

##### 2- tuple 

```python
# Function to demonstrate tuple operations
def tuple_operations():
    my_tuple = (1, 2, "hello", 4, "world")
    # Tuples are immutable, so we can't directly append/remove/update.
    # But we can convert to a list, modify, then convert back.
    temp_list = list(my_tuple)
    temp_list.append("new_item")
    temp_list.remove("hello")
    temp_list[1] = 99
    my_tuple = tuple(temp_list)

    print("Modified Tuple:", my_tuple)
tuple_operations()
```

##### 3- Set

```python
# Function to demonstrate set operations
def set_operations():
    my_set = {1, 2, "hello", 4, "world"}


    # Add (similar to append)
    my_set.add("new_item")
    print("After add:", my_set)

    # Remove
    my_set.remove("hello")
    print("After remove:", my_set)

    # Update (sets don’t have indexes, but we can remove and add)
    my_set.discard(2)
    my_set.add(99)
    print("After update-like operation:", my_set)
set_operations()
```
