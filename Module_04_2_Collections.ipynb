##### Python Collections (Arrays)
* There are four collection data types in the Python programming language:

* List is a collection which is ordered and changeable. Allows duplicate members. []
* Tuple is a collection which is ordered and unchangeable. Allows duplicate members. ()
* Set is a collection which is unordered and unindexed. No duplicate members. {}
* Dictionary is a collection which is unordered, changeable and indexed. No duplicate members. {}

#### List

```python
fruits = ['Apple', 'Banana', 'Mango', 'Pineapple', 'Grapes']
```

####  1. append(): Add an element at the end

```python
fruits.append('Orange')
fruits.append('Guava')
print("After append:", fruits)
```

##### 2. pop(): Remove and return the last element (or element at given index)

```python

popped_item = fruits.pop()  # Removes 'Orange'
print("After pop:", fruits)
print("Popped item:", popped_item)
```

```python
fruits.insert(2, 'Strawberry')  # At index 2
fruits.insert(2, 'Kiwi')
fruits.insert(4, 'Banana')
print("After insert:",fruits)
```

#### 4. remove(): Remove the first occurrence of a value

```python
fruits.remove('Banana')
print("After remove:",fruits)
```

#### 5. sort(): Sort the list in ascending order

```python
fruits.sort()
print("After sort:", fruits)
```

#### 6. reverse(): Reverse the current order of the list

```python
fruits.reverse()
print("After reverse:",fruits)
```

#### 7. clear(): Remove all elements from the list

```python
fruits.clear()
print("After clear:", fruits)
```

```python
#### 8. extend(): Add multiple elements to the list
```

#### Indexing

#### Positive Indexing:
#### Use square brackets [] to access elements.

```python
    my_list = [10, 20, 30, 40, 50]
    print(my_list[0])  # Output: 10 (first element)
    print(my_list[2])  # Output: 30 (third element)
```

#### Negative Indexing:
#### Negative indices count from the end of the sequence. -1 refers to the last element, -2 to the second-to-last, and so on.

```python
    my_list = [10, 20, 30, 40, 50]
    print(my_list[-1])  # Output: 50 (last element)
    print(my_list[-3])  # Output: 30 (third from last)
```

#### Slicing
#### Slicking allows us to extract a part of the string. We can specify a start index, end index, and step size. The general format for slicing is:
#### string[start : end : step]
* start : We provide the starting index.
* end : We provide the end index(this is not included in substring).
* step : It is an optional argument that determines the increment between each index for slicing.


```python
    my_list = [10, 20, 30, 40, 50]
    print(my_list[1:4])  # Output: [20, 30, 40] (elements at indices 1, 2, 3)
    print(my_list[2:])  # Output: [30, 40, 50] (elements from index 2 to the end)
    print(my_list[:3])  # Output: [10, 20, 30] (elements from beginning up to index 3)
    print(my_list[:])  # Output: [10, 20, 30, 40, 50] (entire list)
    print(my_list)
```

#### 3. Set
* A set is a collection which is unordered and unindexed. In Python, sets are written with curly brackets.

```python
object = {1,2,1,1,1,2,2,2,2,4,5,5,5,'gaurav',5,6,'monu',6,6,6,7,7,7}
print(object)
```

```python
set1 = {1, 2, 4, 5, 'Ram', 6, 'Shyam', 7}
```

#### 1. add() – Add a single element to the set

```python
set1.add('Ravi')
print("After add:", set1)
```

#### 2. remove() – Remove a specific element; raises KeyError if not found

```python
set1.remove('Ram')
print("After remove:", set1)
```

#### 3. discard() – Remove a specific element; does not raise an error if not found

```python
set1.discard('Shyam')
set1.discard('NotExist')  # No error
print("After discard:", set1)
```

#### 4. pop() – Remove and return a random element (since sets are unordered)

```python
popped_item = set1.pop()
print("After pop:", set1)
print("Popped item:", popped_item)
```

####  5. clear() – Remove all elements

```python
set1.clear()
print("After clear:", set1)  # Output: set()
```

#### 6.union() – Combines both sets (removes duplicates)

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union_set = set1.union(set2)
print("Union:", union_set)
# Output: {1, 2, 3, 4, 5, 6, 7, 8}
```

#### 7. intersection() – Common elements between both sets

```python
intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)
# Output: {4, 5}
```

#### 8. difference() – Elements in set1 but not in set2

```python
difference_set = set1.difference(set2)
print("Difference (set1 - set2):", difference_set)
difference_set2 = set2.difference(set1)
print("Difference default set2 (set1 - set2):", difference_set2)
# Output: {1, 2, 3}
```

#### 4. Dictionary

* A dictionary is a collection which is unordered, changeable and indexed. In Python dictionaries are written with curly brackets, and they have keys and values.
* dict = {"key1":"value1","key2":"value2",}
* dict

```python
# Create a dictionary with keys and their respective prices as values
fruit_prices = {
    "apple": 50,
    "banana": 20,
    "cherry": 100
}
```

#### 1. get() – Get the value for a key (returns None if key not found)

```python
print("Price of apple:", fruit_prices.get("apple"))     # Output: 50
print("Price of mango:", fruit_prices.get("mango"))     # Output: None
print("Price of mango with default:", fruit_prices.get("mango", 0))  # Output: 0
```

#### 2. keys() – Return all keys

```python
print("Keys:", fruit_prices.keys())
# Output: dict_keys(['apple', 'banana', 'cherry'])
```

#### 3. values() – Return all values

```python
print("Values:", fruit_prices.values())
# Output: dict_values([50, 20, 100])
```

####  4. items() – Return all key-value pairs

```python
print("Items:", fruit_prices.items())
print(fruit_prices)
fruit_prices.update({"apple": 60, "mango": 70, "banana":30})
print(fruit_prices)
# Output: dict_items([('apple', 50), ('banana', 20), ('cherry', 100)])
```

#### 5. pop() – Remove a key and return its value

```python
price = fruit_prices.pop("banana")
print("After pop:", fruit_prices)
print("Popped value:", price)
# Output: {'apple': 50, 'cherry': 100}
```

#### 6. update() – Update existing key or add a new one

```python
fruit_prices.update({"apple": 60, "mango": 70})
print("After update:", fruit_prices)
# Output: {'apple': 60, 'cherry': 100, 'mango': 70}
```

#### 7. clear() – Remove all items

```python
fruit_prices.clear()
print("After clear:", fruit_prices)
# Output: {}
```
