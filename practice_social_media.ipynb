Q1: Total number of posts by each user

Display username and how many posts they made.

Hint: JOIN users to posts, then GROUP BY username with COUNT(*).

```sql
select u.username , count(post_id) as posts
from users u 
left join posts p -- includes 0 number of post 
on u.user_id = p.user_id
group by u.username;



```

```sql
select u.username, count(like_id) as Total_likes
from users u
join likes l 
on u.user_id = l.user_id
group by u.username
```

```sql
SELECT post_id, COUNT(*) AS total_likes
FROM likes
GROUP BY post_id
ORDER BY total_likes DESC;
```

```sql
--No join is needed, because likes already has post_id.

---If you also want posts with 0 likes (6, 7, 8), start from posts:

SELECT p.post_id, COUNT(l.like_id) AS total_likes
FROM posts p
LEFT JOIN likes l ON p.post_id = l.post_id
GROUP BY p.post_id
ORDER BY total_likes DESC;
```

```sql
--If you want likes received per username (a useful variation), go through posts:


SELECT u.username, COUNT(l.like_id) AS likes_received
FROM users u
JOIN posts p ON u.user_id = p.user_id
JOIN likes l ON p.post_id = l.post_id
GROUP BY u.username
ORDER BY likes_received DESC;
```
